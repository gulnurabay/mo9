```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

```

```python
df = pd.read_csv('market_cl.csv')
```

```python
df
```

```python
df.info()
```

```python
spending_columns = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

df['Total_Spending'] = df[spending_columns].sum(axis=1)
```

```python
df['Income'] = df['Income'].fillna(df['Income'].median())
df = df[df['Income'] < 200000].copy()

print(df[['Income', 'Total_Spending']].head())
print("Жаңа өлшемі:", df.shape)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.hist(df['Income'], bins=20, edgecolor='black')
plt.title('Customer Income')
plt.xlabel('Income')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 2)
plt.hist(df['Total_Spending'], bins=20, edgecolor='black')
plt.title('Total Spending')
plt.xlabel('Total Spending')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['Income'],
    df['Total_Spending'],
    alpha=0.6
)

plt.xlabel('Income')
plt.ylabel('Total Spending')
plt.title('Customer Income vs Total Spending')
plt.show()
```

```python
X = df[['Income', 'Total_Spending']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia = []
silhouette_scores = []
k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    inertia.append(model.inertia_)

    silhouette_scores.append(
        silhouette_score(X_scaled, labels)
    )
```

```python
plt.figure(figsize=(8, 5))

plt.plot(k_values, inertia, marker='o')

plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(k_values, silhouette_scores, marker='o')

plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Method')

plt.show()
```

```python
for k, score in zip(k_values, silhouette_scores):
    print(f'K = {k}, Silhouette Score = {score:.3f}')
```

```python
kmeans_scaled = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

labels_scaled = kmeans_scaled.fit_predict(X_scaled)

score_after = silhouette_score(X_scaled, labels_scaled)

print('Silhouette Score после нормализации:',
      round(score_after, 3))
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_scaled)

df['Cluster'] = labels

print(df[['Income', 'Total_Spending', 'Cluster']].head(10))
```

```python
centers = scaler.inverse_transform(kmeans.cluster_centers_)

print("Кластер орталықтары:")
print(centers)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['Income'],
    df['Total_Spending'],
    c=labels,
    cmap='viridis',
    s=50
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    color='red',
    marker='X',
    s=200,
    label='Cluster Centers'
)

plt.xlabel('Income')
plt.ylabel('Total Spending')
plt.title('K-Means Customer Clustering')
plt.legend()
plt.show()
```

```python
cluster_analysis = df.groupby('Cluster').agg({
    'Income': 'mean',
    'Total_Spending': 'mean',
    'ID': 'count'
})

cluster_analysis = cluster_analysis.rename(
    columns={'ID': 'Customer_Count'}
)

print(cluster_analysis)
```
